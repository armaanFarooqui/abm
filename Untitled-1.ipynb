```python

import pyttsx3, sys, os

SCRIPT_PATH = r"C:\Users\jahee\Downloads\dutch_irregulars_script_inf-pres-perf.txt"
OUT_PATH = r"C:\Users\jahee\Downloads\dutch_irregulars_inf-pres-perf.wav"

def pick_dutch_female(engine):
    dutch_voice_id = None
    for v in engine.getProperty('voices'):
        lang = getattr(v, 'languages', [b''])[0]
        lang_str = ""
        if isinstance(lang, bytes):
            try:
                lang_str = lang.decode(errors='ignore').lower()
            except:
                lang_str = ""
        else:
            lang_str = str(lang).lower()
        name_lower = v.name.lower()
        if ("nl" in lang_str or "dutch" in name_lower or "nl_" in name_lower) and ("female" in name_lower or "vrouw" in name_lower or "fem" in name_lower):
            return v.id
    # fallback to any Dutch voice
    for v in engine.getProperty('voices'):
        lang = getattr(v, 'languages', [b''])[0]
        lang_str = ""
        if isinstance(lang, bytes):
            try:
                lang_str = lang.decode(errors='ignore').lower()
            except:
                lang_str = ""
        else:
            lang_str = str(lang).lower()
        name_lower = v.name.lower()
        if ("nl" in lang_str) or ("dutch" in name_lower) or ("nl_" in name_lower):
            return v.id
    return None

def main():
    if not os.path.exists(SCRIPT_PATH):
        print("Could not find script text:", SCRIPT_PATH)
        sys.exit(1)
    with open(SCRIPT_PATH, "r", encoding="utf-8") as f:
        text = f.read()
    engine = pyttsx3.init()
    voice_id = pick_dutch_female(engine)
    if voice_id:
        engine.setProperty('voice', voice_id)
    # slightly slower for clarity
    rate = engine.getProperty('rate')
    if isinstance(rate, int):
        engine.setProperty('rate', int(rate * 0.9))
    # Save to WAV
    engine.save_to_file(text, OUT_PATH)
    engine.runAndWait()
    print("Saved:", OUT_PATH)

if __name__ == "__main__":
    main()
```
