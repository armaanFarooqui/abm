### Total Bites (By Group)

```python
import pandas as pd
import matplotlib.pyplot as plt

# Input path
csv_path = r"C:\Users\arman\OneDrive - University of Twente\assignment_spatio\data\outputs\csv\total_group.csv"

# Output path
output_path = r"C:\Users\arman\OneDrive - University of Twente\assignment_spatio\data\outputs\plots\total_group.png"

# Loading the CSV file
data = pd.read_csv(csv_path)

# Checking if the CSV has at least two columns
if data.shape[1] < 2:
    raise ValueError("CSV file must have at least two columns (one for x-axis and at least one for y-axis).")

# Extracting x-axis (first column) and y-axis columns (remaining columns)
x = data.iloc[:, 0]
y_columns = data.columns[1:]

# Creating the plot
plt.figure(figsize=(10, 6))
for col in y_columns:
    plt.plot(x, data[col], label=col)

# Adding labels and title
plt.xlabel("Ticks")
plt.ylabel("Values")
plt.title("Total Bites (By Group)")
plt.legend()
plt.grid(True)
plt.tight_layout()

# Saving and displaying the figure
plt.savefig(output_path)
plt.show()
plt.close()

print(f"Plot saved to {output_path}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Input CSV path
csv_path = r"C:\Users\arman\OneDrive - University of Twente\assignment_spatio\data\outputs\csv\new_group.csv"

# Output paths for plots
output_sum_path = r"C:\Users\arman\OneDrive - University of Twente\assignment_spatio\data\outputs\plots\new_group_monthly_sum.png"
output_avg_path = r"C:\Users\arman\OneDrive - University of Twente\assignment_spatio\data\outputs\plots\new_group_monthly_avg.png"

# Load the CSV
df = pd.read_csv(csv_path)

# Rename the first column to "tick" for consistency
df.rename(columns={df.columns[0]: "tick"}, inplace=True)

# Create a 'month' column from tick (assuming 1 tick = 1 day, 30-day months)
df["month"] = ((df["tick"] % 365) // 30 + 1).astype(int)

# Compute monthly statistics
monthly_sum = df.groupby("month").sum(numeric_only=True)
monthly_avg = df.groupby("month").mean(numeric_only=True)

# Plot monthly SUM
plt.figure(figsize=(10, 6))
for col in monthly_sum.columns:
    plt.plot(monthly_sum.index, monthly_sum[col], marker='o', label=col)
plt.title("Monthly Total New Bites by Group")
plt.xlabel("Month")
plt.ylabel("Total Bites")
plt.xticks(range(1, 13))
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(output_sum_path)
plt.show()

# Plot monthly AVERAGE
plt.figure(figsize=(10, 6))
for col in monthly_avg.columns:
    plt.plot(monthly_avg.index, monthly_avg[col], marker='o', label=col)
plt.title("Monthly Average Daily New Bites by Group")
plt.xlabel("Month")
plt.ylabel("Average Daily Bites")
plt.xticks(range(1, 13))
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(output_avg_path)
plt.show()

print(f"Plots saved to:\n{output_sum_path}\n{output_avg_path}")
```
