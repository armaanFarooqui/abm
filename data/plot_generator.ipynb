### Total Bites

```python
import pandas as pd
import matplotlib.pyplot as plt

# Input path
csv_path = r"C:\Users\arman\OneDrive - University of Twente\assignment_spatio\data\outputs\csv\bite_count.csv"

# Output path
output_path = r"C:\Users\arman\OneDrive - University of Twente\assignment_spatio\data\outputs\plots\total_bites.png"

# Loading the CSV file
data = pd.read_csv(csv_path)

# Checking if the CSV has at least two columns
if data.shape[1] < 2:
    raise ValueError("CSV file must have at least two columns (one for x-axis and at least one for y-axis).")

# Extracting x-axis (first column) and y-axis columns (remaining columns)
x = data.iloc[:, 0]
y_columns = data.columns[1:]

# Creating the plot
plt.figure(figsize=(10, 6))
for col in y_columns:
    plt.plot(x, data[col], label=col)

# Adding labels and title
plt.xlabel("Ticks")
plt.ylabel("Values")
plt.title("Total Bites")
plt.legend()
plt.grid(True)
plt.tight_layout()

# Saving and displaying the figure
plt.savefig(output_path)
plt.show()
plt.close()

print(f"Plot saved to {output_path}")
```
