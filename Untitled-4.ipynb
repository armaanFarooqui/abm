```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# ==============================
# RAW RESULTS (Netherlands-wide)
# ==============================

data = [
    ("Shapefile", "format", 2234.89, 37.2721, 197.9704),
    ("FlatGeobuf", "format", 1108.67, 35.4475, 191.1127),
    ("GeoParquet (default)", "format", 1057.45, 12.9006, 13.2871),

    ("GeoParquet GZIP (L1)", "compression", 894.91, 20.5072, 490.9820),
    ("GeoParquet GZIP (L9)", "compression", 779.97, 17.5195, 90.1724),
    ("GeoParquet ZSTD (L1)", "compression", 968.22, 13.7165, 12.9536),
    ("GeoParquet ZSTD (L9)", "compression", 904.39, 13.0340, 34.7934),
    ("GeoParquet ZSTD (L15)", "compression", 904.90, 13.4834, 139.5469),
    ("GeoParquet ZSTD (L22)", "compression", 675.70, 12.0518, 336.8660),
    ("GeoParquet BROTLI (L5)", "compression", 660.99, 16.7893, 124.4306),
    ("GeoParquet BROTLI (L11)", "compression", 652.96, 17.2537, 2518.8486),
]

df = pd.DataFrame(
    data,
    columns=["Label", "Category", "Size_MB", "Read_s", "Write_s"]
)

# Output directory
fig_dir = "figures"
os.makedirs(fig_dir, exist_ok=True)


# ==============================
# HELPER FUNCTION
# ==============================

def save_bar(df, x, y, title, ylabel, filename):
    plt.figure(figsize=(12, 5))
    plt.bar(df[x], df[y])
    plt.xticks(rotation=45, ha="right")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.tight_layout()
    plt.savefig(os.path.join(fig_dir, f"{filename}.png"), dpi=300)
    plt.savefig(os.path.join(fig_dir, f"{filename}.pdf"))
    plt.show()


# ==============================
# GROUP A: FORMAT COMPARISON
# ==============================

df_format = df[df["Category"] == "format"]

save_bar(
    df_format,
    x="Label",
    y="Size_MB",
    title="Netherlands-wide Vector Dataset – File Size by Format",
    ylabel="File size (MB)",
    filename="A_format_size"
)

save_bar(
    df_format,
    x="Label",
    y="Read_s",
    title="Netherlands-wide Vector Dataset – Read Time by Format",
    ylabel="Read time (s)",
    filename="A_format_read"
)

save_bar(
    df_format,
    x="Label",
    y="Write_s",
    title="Netherlands-wide Vector Dataset – Write Time by Format",
    ylabel="Write time (s)",
    filename="A_format_write"
)


# ==============================
# GROUP B: COMPRESSION ANALYSIS
# ==============================

df_comp = df[df["Category"] == "compression"]

save_bar(
    df_comp,
    x="Label",
    y="Size_MB",
    title="Netherlands-wide Vector Dataset – Effect of Compression on File Size",
    ylabel="File size (MB)",
    filename="B_compression_size"
)

save_bar(
    df_comp,
    x="Label",
    y="Read_s",
    title="Netherlands-wide Vector Dataset – Effect of Compression on Read Time",
    ylabel="Read time (s)",
    filename="B_compression_read"
)

save_bar(
    df_comp,
    x="Label",
    y="Write_s",
    title="Netherlands-wide Vector Dataset – Effect of Compression on Write Time",
    ylabel="Write time (s)",
    filename="B_compression_write"
)

print(f"Figures saved to '{fig_dir}/'")
```
